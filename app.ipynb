```python
import pandas as pd 
import numpy as np
import ast
```

```python
movies = pd.read_csv("tmdb_5000_movies.csv")
credits = pd.read_csv("tmdb_5000_credits.csv")
```

```python
movies.head()
```

```python
credits.head()
```

```python
movies = movies.merge(credits, on="title")
```

```python
movies.info()
```

```python
# columns we need for this project

movies = movies[["movie_id", "title", "overview", "genres", "keywords", "cast", "crew"]]
```

```python
# preprocessing

movies.isnull().sum()
```

```python
movies.dropna(inplace = True)
```

```python
movies.duplicated().sum()
```

```python
ast.literal_eval(movies["genres"][0])
```

```python
# function to extract names from list of dictionary

def convert(object):
    list = []
    for i in ast.literal_eval(object):
        list.append(i["name"])
        
    return list
```

```python
movies["genres"] = movies["genres"].apply(convert)
```

```python
movies["keywords"] = movies["keywords"].apply(convert)
```

```python
# function to extract top 3 casts

def convert_cast(object):
    list = []
    counter = 1
    for i in ast.literal_eval(object):
        if counter <= 3:
            list.append(i["name"])
            counter+=1
        else:
            break
    return list
```

```python
movies["cast"] = movies["cast"].apply(convert_cast)
```

```python
# function to extract director name

def convert_crew(object):
    list = []
    for i in ast.literal_eval(object):
        if i["job"] == "Director":
            list.append(i["name"])
            break
        
    return list
```

```python
movies["crew"] = movies["crew"].apply(convert_crew)
```

```python
movies["overview"] = movies["overview"].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
# removing all the spaces between words

movies["genres"] = movies["genres"].apply(lambda x:[i.replace(" ", "") for i in x])
movies["keywords"] = movies["keywords"].apply(lambda x:[i.replace(" ", "") for i in x])
movies["cast"] = movies["cast"].apply(lambda x:[i.replace(" ", "") for i in x])
movies["crew"] = movies["crew"].apply(lambda x:[i.replace(" ", "") for i in x])
```

```python
movies["tags"] = movies["overview"] + movies["genres"] + movies["keywords"] + movies["cast"] + movies["crew"]
```

```python
new_df = movies[["movie_id", "title", "tags"]]
```

```python
new_df["tags"] = new_df["tags"].apply(lambda x:" ".join(x))
```

```python
new_df["tags"] = new_df["tags"].apply(lambda x:x.lower())
```

```python
new_df.head()
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
# function to perform stemming

def stem(text):
    list = []
    for i in text.split():
        list.append(ps.stem(i))
    
    return " ".join(list)
```

```python
new_df["tags"] = new_df["tags"].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors = cv.fit_transform(new_df["tags"]).toarray()
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
# function to recommend 5 movies

def recommend(movie):
    movie_index = new_df[new_df["title"] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Avatar")
```
